Null Hypothesis (H₀): There is no significant difference in the mean diameter of cutlets between Unit A and Unit B. (i.e., μA = μB)


Alternative Hypothesis (H₁): There is a significant difference in the mean diameter of cutlets between Unit A and Unit B. (i.e., μA ≠ μB)

Significance Level:

α = 0.05 (5% significance level)

```python
# Extract and clean data
unit_a = df['Unit A'].dropna().astype(float)
unit_b = df['Unit B'].dropna().astype(float)
```

```python
# Check for normality using the Shapiro-Wilk test
stat_a, p_a = stats.shapiro(unit_a)
stat_b, p_b = stats.shapiro(unit_b)

print(f"Shapiro-Wilk test for Unit A: p-value = {p_a}")
print(f"Shapiro-Wilk test for Unit B: p-value = {p_b}")
```

Unit A: p-value = 0.31998

Unit B: p-value = 0.52251

Since both p-values are greater than 0.05, we fail to reject the null hypothesis. This means that the data for both Unit A and Unit B follows a normal distribution.

```python
# variance test for equality using levene test
stat, p_value = stats.levene(unit_a, unit_b)
print(f"Levene's test p-value = {p_value}")
```

Null Hypothesis (H₀): The variances of Unit A and Unit B are equal.
    
Alternative Hypothesis (H₁): The variances of Unit A and Unit B are not equal.

p-value = 0.4176 (which is greater than 0.05)

Fail to reject H₀ → We assume equal variances between Unit A and Unit B.

```python
# If variances are equal, perform a Two-sample t-test (equal variance)
t_stat, p_value = stats.ttest_ind(unit_a, unit_b, equal_var=True)

print(f"T-statistic: {t_stat}, p-value: {p_value}")
```

Interpretation:
Result= p-value=0.4722 ,which is greater than 0.05

H0: There is no significant difference in the mean diameter of cutlets between Unit A and Unit B
    
H1: There is a significant difference in the mean diameter of cutlets between Unit A and Unit B
    
Since the p-value (0.4722) is greater than 0.05, we fail to reject the null hypothesis (H₀)

Conclusion:  There is no significant difference in the mean diameter of cutlets between Unit A and Unit B.
