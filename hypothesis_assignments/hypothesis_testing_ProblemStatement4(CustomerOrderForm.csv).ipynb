# Problem Statement 4-->

Telecall uses 4 centers around the globe to process customer order forms. They audit a certain % of the customer order forms. Any error in order form renders it defective and must be reworked before processing. The manager wants to check whether the defective % varies by center. Please analyze the data at 5% significance level and help the manager draw appropriate inferences

File: Customer OrderForm.csv

# Business Objectives-->


# Constraints-->


```python
#load a required libraries
import pandas as pd
import numpy as np
import scipy 
from scipy import stats
```

```python
#Load a dataset
df=pd.read_csv("CustomerOrderForm.csv")
```

```python
#print 1st 5 rows of dataset
df.head(5)
```

```python
#print information of dataset
df.info()
```

```python
#print shape of dataset
df.shape
```

```python
# Convert categorical values into binary (Error Free = 0, Defective = 1)
df.replace({"Error Free": 0, "Defective": 1}, inplace=True)
```

```python
df.columns
```

```python
df.head(1)
```

```python
# Create a contingency table (counts of defective and error-free)
contingency_table = df.apply(pd.Series.value_counts).T
```

```python
print("\nContingency Table (Defective vs. Error-Free):")
print(contingency_table)
```

```python
# Perform Chi-Square Test for Independence
chi2_stat, p_value, dof, expected = stats.chi2_contingency(contingency_table)
```

```python
# Display results
print("\nChi-Square Test Results:")
print(f"Chi-Square Statistic = {chi2_stat:.4f}")
print(f"p-value = {p_value:.4f}")
print(f"Degrees of Freedom = {dof}")
print("\nExpected Frequencies (if no significant difference):")
print(expected)
```

```python
# Interpretation based on significance level alpha = 0.05
alpha = 0.05  
if p_value < alpha:
    print("\nConclusion: Reject H0. The defective % varies significantly by center.")
else:
    print("\nConclusion: Fail to reject H0. No significant difference in defective % among centers.")
```

Interpretation of Chi-Square Test Results:

Result: p-value = 0.2771

Interpretability:
H0 (Null Hypothesis): The proportion of defective order forms is the same across all four centers.

H1 (Alternative Hypothesis): The proportion of defective order forms differs among the four centers.

Since the p-value (0.2771) > 0.05, we fail to reject the null hypothesis.

Conclusion:
There is no significant difference in defect rates among the four centers.
The defects appear to be randomly distributed rather than associated with a specific center.
