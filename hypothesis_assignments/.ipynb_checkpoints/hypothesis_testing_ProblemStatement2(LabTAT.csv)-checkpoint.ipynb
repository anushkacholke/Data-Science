# Problem Statement 2:-->

A hospital wants to determine whether there is any difference in the average Turn Around Time (TAT) of reports of the laboratories on their preferred list. They collected a random sample and recorded TAT for reports of 4 laboratories. TAT is defined as sample collected to report dispatch.
Analyze the data and determine whether there is any difference in average TAT among the different laboratories at 5% significance level.

File: LabTAT.csv

# Business Objectives-->

Ensure efficiency in report generation

compare laboratories performance

improve patient experience-->faster report means quicker treatment desicion

# Constraints-->

data normality assumption-->data must be normally distributed

varience assumption-->the variance across the four groups should be approximately equal

```python
#import required libraries
import pandas as pd
import numpy as np
import scipy 
from scipy import stats
```

```python
#load the dataset
df=pd.read_csv("lab_tat_updated.csv")
```

```python
#print the 1st 5 rows of dataset
df.head(5)
```

```python
#print information of dataset
df.info()
```

```python
#print description of dataset
df.describe()
```

```python
#dataset columns
df.columns=["lab_1","lab_2","lab_3","lab_4"]
df.columns
```

```python
#shape of dataset
df.shape
```

Assumption of normality testing:

H0 (Null Hypothesis): Data follows a normal distribution.

H1 (Alternative Hypothesis): Data does not follow a normal distribution.

```python
#Normality test
print("Lab 1 normality: ",stats.shapiro(df.lab_1))
print("Lab 2 normality: ",stats.shapiro(df.lab_2))
print("Lab 3 normality: ",stats.shapiro(df.lab_3))
```

Lab 1 p-value = 0.4232 > 0.05--> Fail to reject H0--> Data is normal 

Lab 2 p-value = 0.8637 > 0.05--> Fail to reject H0--> Data is normal 

Lab 3 p-value = 0.0655 > 0.05--> Fail to reject H0--> Data is normal 

Since all p-values are greater than 0.05, we fail to reject the null hypothesis.

The data follows a normal distribution for all laboratories.

```python
#variance test
levene_test=scipy.stats.levene(df.lab_1,df.lab_2,df.lab_3)
print("Levene Test (Variance: ",levene_test)
```

H0-->Variances are equal across all laboratories

H1-->Variances are not equal across laboratories

p-value = 0.2163 > 0.05--> Fail to reject H0--> Variances are equal

```python
#ANOVA test
anova_result = stats.f_oneway(df.lab_1, df.lab_2, df.lab_3) 
print("One_way ANOVA:", anova_result)
```

Result p-value=1.0799

Interpretation:

H0: The mean Turn Around Time (TAT) is the same for all laboratories.

H1: At least one laboratory has a significantly different mean TAT.

Since p-value < 0.05, we reject the null hypothesis (H0).

At least one laboratory has a significantly different mean TAT compared to others.
