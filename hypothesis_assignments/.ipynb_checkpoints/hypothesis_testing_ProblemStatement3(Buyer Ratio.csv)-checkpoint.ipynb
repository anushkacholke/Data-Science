# Problem Statement 3-->
Sales of products in four different regions is tabulated for males and females. Find if male-female buyer rations are similar across regions.

Buyer Ratio.csv

# Business Objectives:-->

To determine whether the ratio of male to female buyers is similar across four different regions (East, West, North, South)

This will help in understanding buyer demographics and assist in regional marketing strategies.

# Constraints:-->
The sample data must be representative of the population.

The chi-square test assumes that expected frequencies should not be too low (generally >5).

```python
#import required libraries
import numpy as np
import pandas as pd
import scipy.stats as stats
```

```python
#load the dataset
df=pd.read_csv("BuyerRatio.csv")
```

```python
#description of dataset
df.info()
```

```python
#shape of dataset
df.shape
```

```python
#describe the dataset
df.describe()
```

```python
df.columns
```

```python
#convert DataFrame to Numpy Array for Analysis
observed = df.iloc[:, 1:].values  # Assuming first column contains labels (Males, Females)
print("Observed Frequencies:\n", observed)
```

```python
#perform the Chi-Square test
chi2_stat, p_value, dof, expected = stats.chi2_contingency(observed)
```

```python
# Print results
print("\nChi-Square Statistic:", chi2_stat)
print("p-value:", p_value)
print("Degrees of Freedom:", dof)
print("\nExpected Frequencies (if male-female buyer ratios were similar):\n", pd.DataFrame(expected, index=df.iloc[:, 0], columns=df.columns[1:]))
```

Result: p-value = 0.6603

Interpretability:
 
H0: Male-female buyer ratios are the same across all regions.
  
H1: Male-female buyer ratios are different across regions.

Since the p-value (0.6603) is greater than 0.05, we accept the null hypothesis.

Conclusion: Male-female buyer ratios are similar across all regions.
